```python
A = np.array([[1, 2], [3, 4]])
n = 2

# 矩阵的乘方
A_n = np.linalg.matrix_power(A, n)
np.mat(A)**n
print("矩阵的乘方 A^2:\n", A_n)
np.mat(A)**n


A**2
np.power(A,2)
```
