```python
import numpy as np

# 定义旋转角度 (以弧度为单位)
theta = np.pi / 4  # 45度

# 生成二维旋转矩阵
Q = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])

# 计算 Q 的转置
Q_T = Q.T

# 计算 Q^T Q
result = np.dot(Q_T, Q)

print("Q:")
print(Q)
print("\nQ^T:")
print(Q_T)
print("\nQ^T Q:")
print(result)
```
